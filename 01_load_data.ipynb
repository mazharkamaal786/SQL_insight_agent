```python
import pandas as pd 
import numpy as np


np.random.seed(42)
n = 500

employees = {
    101: ("Ravi Kumar", "North"),
    102: ("Priya Singh", "North"),
    103: ("Amit Verma", "South"),
    104: ("Sneha Gupta", "East"),
    105: ("Imran Khan", "West"),
    106: ("Neha Sharma", "West"),
}

emp_ids = np.random.choice(list(employees.keys()), n, p=[0.25, 0.15, 0.10, 0.20, 0.20, 0.10])

df = pd.DataFrame({
    "sale_id": range(1, n + 1),
    "emp_id": emp_ids,
    "emp_name": [employees[i][0] for i in emp_ids],
    "region": [employees[i][1] for i in emp_ids],
    "sale_date": (pd.to_datetime("2025-01-01")
                  + pd.to_timedelta(np.random.randint(0, 365, n), unit="D")).date,
    "amount": np.random.randint(500, 20000, n),
})

df.to_csv("sales.csv", index=False)
df.head()
```
